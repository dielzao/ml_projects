# Exploratory data analysis

```python
import pandas as pd
import numpy as np
```

## Import the data

```python
df = pd.read_csv(r"C:\Users\diego\Documents\ml_projects\iris\IRIS.csv")
```

```python
# Print preview of the dataset
print(df.head())
```

## Plot information from the data
- Number of observations
- Number of features
- Features names
- Data type
- Label column
- Units of measurement
- Number of unique values in categorical variables

```python
print("Number of observations:", df.shape[0])
print("Number of features:", df.shape[1])
print("Feature names:", df.columns.tolist()[:-1])
print("Label name:", df.columns.tolist()[-1])
print("Class distribution:\n", df[df.columns[-1]].value_counts())
print("Data types:", df.dtypes)
```

## Data quality
- Missing values
- Duplicated observations
- Impossible values
- Constant features
- Invalid data types

```python
print("Missing values:", df.isnull().sum())
print("Duplicated observations:", df.duplicated().sum())
print("Constant features:", df.nunique()[df.nunique() == 1].index.tolist())
```

```python
# Remove duplicated observations if any
if df.duplicated().any():
    df = df.drop_duplicates()
    print("Duplicated observations removed. New shape:", df.shape)


print("\nClass distribution:\n", df[df.columns[-1]].value_counts())
```

## Data statistics
- Mean
- Median
- Standard deviation
- Minimum
- Maximum
- Quartiles
- Possible skewness
- Correlation matrix

```python
print("statistics summary:\n", df.describe())
print("\nCorrelation matrix:\n", df.iloc[:, :-1].corr())
print("\nVariance of features:\n",df.iloc[:, :-1].var())
```
